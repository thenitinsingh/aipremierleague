```python
# Standardizing Venue Names
venue_map = {
    "M.Chinnaswamy Stadium": "M Chinnaswamy Stadium",
    "Sardar Patel Stadium": "Narendra Modi Stadium",
    "Punjab Cricket Association IS Bindra Stadium": "Punjab Cricket Association Stadium",
    "Feroz Shah Kotla": "Arun Jaitley Stadium",
    "Zayed Cricket Stadium": "Sheikh Zayed Stadium",
    "De Beers Diamond Oval": "Kimberley Oval",
    "Rajiv Gandhi International Stadium": "Uppal Stadium",
    "Dr Y.S. Rajasekhara Reddy ACA-VDCA Cricket Stadium": "ACA-VDCA Stadium",
    "Shaheed Veer Narayan Singh International Stadium": "Raipur International Cricket Stadium",
    "JSCA International Stadium Complex": "Ranchi Stadium",
    "Barsapara Cricket Stadium": "Assam Cricket Association Stadium"
}
data["Venue"] = data["Venue"].replace(venue_map)
```

```python
data['Batter'].unique()
```

```python
# Standardizing Player Names
data["Batter"] = data["Batter"].str.strip().str.title()
data["Non Striker"] = data["Non Striker"].str.strip().str.title()
data["Bowler"] = data["Bowler"].str.strip().str.title()
data["Player Out"] = data["Player Out"].str.strip().str.title()
```

```python
data.dtypes #to display data-type of each column
```

```python
# Convert Ball Rebowled to boolean
data['Ball Rebowled'] = data['Ball Rebowled'].astype(bool)
# Convert Runs to Get to integer
data['Runs to Get'] = pd.to_numeric(data['Runs to Get'], errors='coerce').fillna(0).astype(int)
# Convert Player Out Runs to integer
data['Player Out Runs'] = pd.to_numeric(data['Player Out Runs'], errors='coerce').fillna(0).astype(int)
# Convert Player Out Balls Faced to integer
data['Player Out Balls Faced'] = pd.to_numeric(data['Player Out Balls Faced'], errors='coerce').fillna(0).astype(int)
```

```python
#Convert date column to datetime
data['Date'] = pd.to_datetime(
    data['Date'],
    errors='coerce'
)
# Extract year
data['Year'] = data['Date'].dt.year

# Check invalid dates
print(data['Date'].isnull().sum())
```

```python
# Matches won by each team
matches_won = data.groupby('Winner')['Match ID'].nunique()
# Matches played by each team
teams1 = data[['Match ID', 'Bat First']].rename(columns={'Bat First': 'Team'})
teams2 = data[['Match ID', 'Bat Second']].rename(columns={'Bat Second': 'Team'})
all_teams = pd.concat([teams1, teams2]).drop_duplicates()
matches_played = all_teams.groupby('Team')['Match ID'].nunique()
# Win rate
win_rate = (matches_won / matches_played).fillna(0)
# Map win rate to dataframe
data['bat_first_win_rate'] = data['Bat First'].map(win_rate)
data['bat_second_win_rate'] = data['Bat Second'].map(win_rate)
print(data[['bat_first_win_rate', 'bat_second_win_rate']].head())
```

```python
batter_yearly_runs = data.groupby(['Year', 'Batter'])['Batter Runs'].sum().reset_index()

# Get the highest run-scorer for each year (Orange Cap winner)
orange_cap_winners = batter_yearly_runs.loc[batter_yearly_runs.groupby('Year')['Batter Runs'].idxmax()]

# Rename columns for clarity
orange_cap_winners = orange_cap_winners.rename(columns={'Batter Runs': 'Total Runs in Season'})

# Sort by year
orange_cap_winners = orange_cap_winners.sort_values('Year')

# Display the results
orange_cap_winners
```

```python
#Top Teams by wins
plt.figure(figsize=(12,6))
data['Winner'].value_counts().plot(kind='bar')
plt.title('Matches Won by Teams')
plt.xlabel('Teams')
plt.ylabel('Wins')
plt.xticks(rotation=45)
plt.savefig(
    '../visuals/top_teams.png',
    bbox_inches='tight'
)
plt.show()
```

```python
#Run Distribution
plt.figure(figsize=(10,5))
sns.histplot(data['Runs From Ball'], bins=10)
plt.title('Distribution of Runs Per Ball')
plt.savefig(
    '../visuals/run_distribution.png',
    bbox_inches='tight'
)
plt.show()
```

```python
#Top Batters by Total Runs
top_batters = (
    data.groupby('Batter')['Batter Runs']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
plt.figure(figsize=(12,6))
top_batters.plot(kind='bar')
plt.title('Top 10 Batters by Runs')
plt.ylabel('Total Runs')
plt.xticks(rotation=45)
plt.savefig(
    '../visuals/top_batters.png',
    bbox_inches='tight'
)
plt.show()
```

```python
#Top Bowlers by Total Wickets
top_bowlers = (
    data[data['Player Out'] != 'None']
    .groupby('Bowler')['Player Out']
    .count()
    .sort_values(ascending=False)
    .head(10)
)
plt.figure(figsize=(12,6))
top_bowlers.plot(kind='bar')
plt.title('Top Bowlers by Wickets')
plt.ylabel('Wickets')
plt.xticks(rotation=45)
plt.savefig(
    '../visuals/top_bowlers.png',
    bbox_inches='tight'
)
plt.show()
```

```python
#Correlation Heatmap
numeric_data = data.select_dtypes(include=['int64', 'float64'])
plt.figure(figsize=(12,8))
sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Feature Correlation Heatmap')
plt.show()
```
